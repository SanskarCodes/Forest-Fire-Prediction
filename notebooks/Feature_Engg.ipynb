```python
import pandas as pd
import numpy as np
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Algerian_forest_fires_dataset_UPDATE.csv',header=1)
```

```python
df.info()
```

```python
df[df.isnull().any(axis=1)]
```

```python
df.loc[:122,'Region'] = 0 #Bejaia Region
df.loc[122:,'Region'] = 1 #Sidi-Bel Abbes Region
```

```python
df.info()
```

```python
df['Region'] = df['Region'].astype(int)
```

```python
df.info()
```

```python
df.head()
```

```python
df=df.dropna().reset_index(drop=True)
```

```python
df.isnull().sum()
```

```python
df.iloc[[122]]
```

```python
df=df.drop(122).reset_index(drop=True)
```

```python
df.columns
```

```python
#there are some spaces in between we have to remove them:-
df.columns=df.columns.str.strip()
df.columns
```

```python
df.head()
```

```python
df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']]=df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']].astype(int)
```

```python
df.info()
```

```python
df[['Rain', 'FFMC',
       'DMC', 'DC', 'ISI', 'BUI', 'FWI']]=df[['Rain', 'FFMC',
       'DMC', 'DC', 'ISI', 'BUI', 'FWI']].astype(float)
```

# EDA

```python
df_copy = df.drop(['day','month','year'],axis = 1)
```

```python
df_copy.head()
```

```python
df_copy['Classes'].value_counts()
```

```python
df_copy['Classes'].unique()
```

```python

df_copy['Classes']=np.where(df_copy['Classes'].str.contains('not fire'),0,1)
```

```python
df_copy['Classes'].dtype
```

```python
df_copy['Classes'].value_counts()
```

```python
plt.style.available
```

```python
plt.style.use('seaborn-v0_8')
df_copy.hist(bins=20,figsize=(20,15))
plt.show()
```

```python
#pie chart :-

plt.style.use('ggplot')
percentage = df_copy['Classes'].value_counts(normalize=True) * 100
plt.figure(figsize=(8, 6))
plt.pie(
    percentage,
    labels=['Fire', 'Not Fire'],
    autopct='%1.1f%%',
    startangle=90,
    shadow=True,
    explode=(0.05, 0),          # optional: highlight first slice
    wedgeprops={'edgecolor': 'white'}
)
plt.title('Pie Chart of Classes')
plt.axis('equal')               
plt.show()
```

```python
#Correlation:-
df_copy.corr()
```

```python
plt.style.use('seaborn-v0_8')
sns.heatmap(df_copy.corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    square=True,
    cbar=True)


plt.title("Correlation Matrix", fontsize=16)
plt.tight_layout()
```
